```python
# Example 7: Correlation Matrix of Features
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


df = pd.DataFrame(np.random.randn(200, 5), columns=list("ABCDE"))

agent = PlotAgent()
agent.set_df(df)

# Initial plot
response = agent.process_message(
    "Compute and display a heatmap of the correlation matrix between the five features."
)
fig = agent.get_figure()
fig.show()

# Follow-up iteration
response = agent.process_message(
    "Mask the upper triangle, annotate each cell with the correlation coefficient (two decimals), and apply a diverging color scale."
)
fig = agent.get_figure()
fig.show()
```
