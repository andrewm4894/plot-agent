```python
# Example 5: Transaction Data
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


df = pd.DataFrame({
    "transaction_id": np.arange(1, 501),
    "item": np.random.choice([f"Item {i}" for i in range(1, 21)], 500),
    "quantity": np.random.poisson(3, 500)
})

agent = PlotAgent()
agent.set_df(df)

# Initial plot
response = agent.process_message(
    "Show a bar chart of the top 10 most frequently purchased items by total quantity."
)
fig = agent.get_figure()
fig.show()

# Follow-up iteration
response = agent.process_message(
    "Reorder bars descending, add data labels on each bar, and give the plot a dark grid background."
)
fig = agent.get_figure()
fig.show()
```
