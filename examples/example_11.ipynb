```python
# Example 11: Clustered Correlation Heatmap
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


# Generate sample financial/stock data with correlations
np.random.seed(42)
n_days = 252  # One trading year

# Create correlated asset returns
base = np.random.randn(n_days)
assets = {
    "AAPL": base * 0.8 + np.random.randn(n_days) * 0.5,
    "MSFT": base * 0.75 + np.random.randn(n_days) * 0.6,
    "GOOGL": base * 0.7 + np.random.randn(n_days) * 0.55,
    "AMZN": base * 0.65 + np.random.randn(n_days) * 0.7,
    "META": base * 0.6 + np.random.randn(n_days) * 0.8,
    "NVDA": np.random.randn(n_days) * 1.2,  # Less correlated
    "JPM": -base * 0.3 + np.random.randn(n_days) * 0.9,  # Negative correlation
    "GS": -base * 0.25 + np.random.randn(n_days) * 0.85,
    "XOM": np.random.randn(n_days) * 0.7,  # Independent
    "CVX": np.random.randn(n_days) * 0.65,
}

df = pd.DataFrame(assets)

# Calculate correlation matrix and reshape for heatmap
corr_matrix = df.corr()
corr_df = corr_matrix.reset_index().melt(id_vars="index", var_name="asset_2", value_name="correlation")
corr_df = corr_df.rename(columns={"index": "asset_1"})

agent = PlotAgent()
agent.set_df(corr_df)

# Create clustered correlation heatmap
response = agent.process_message(
    "Create a clustered heatmap showing the correlation between assets. "
    "Use asset_1 on the y-axis, asset_2 on the x-axis, and correlation for the color. "
    "Apply hierarchical clustering to both axes to group similar assets together. "
    "Use a diverging color scale (red-white-blue) centered at 0."
)
fig = agent.get_figure()
fig.show()

# Refine the visualization
response = agent.process_message(
    "Add correlation values as text annotations on each cell, "
    "title it 'Asset Correlation Matrix (Clustered)', "
    "and add a colorbar title 'Pearson Correlation'."
)
fig = agent.get_figure()
fig.show()
```
