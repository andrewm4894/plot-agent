```python
# Example 10: Sensor Readings Heatmap
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


times = pd.date_range("2025-04-01 00:00", periods=24, freq="H")
sensors = [f"Sensor {i}" for i in range(1, 11)]
df = (
    pd.DataFrame(np.random.poisson(20, (24, 10)), index=times, columns=sensors)
      .reset_index()
      .melt(id_vars="index", var_name="sensor", value_name="count")
      .rename(columns={"index": "timestamp"})
)

agent = PlotAgent()
agent.set_df(df)

# Initial plot
response = agent.process_message(
    "Create a heatmap of sensor counts with timestamps on the y-axis and sensors on the x-axis."
)
fig = agent.get_figure()
fig.show()

# Follow-up iteration
response = agent.process_message(
    "Cluster both axes by similarity, add a colorbar title 'Event Count', and rotate x-axis labels 45°."
)
fig = agent.get_figure()
fig.show()
```
