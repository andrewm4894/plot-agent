```python
# Example 1: Daily Temperature & Precipitation
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


df = pd.DataFrame({
    "date": pd.date_range("2025-01-01", periods=100, freq="D"),
    "avg_temp": np.random.normal(15, 5, 100),
    "precip_mm": np.abs(np.random.normal(3, 2, 100))
})

# Initialize the agent
agent = PlotAgent()
agent.set_df(df)

# Create dual-axis line chart
response = agent.process_message(
    "Create a dual-axis line chart showing daily average temperature on the left y-axis and precipitation (mm) on the right y-axis over time."
)
fig = agent.get_figure()
fig.show()

# Interactively update the plot
response = agent.process_message(
    "Smooth the temperature line with a 14-day rolling average, add shaded confidence bands (±1σ), and title the chart 'Winter Weather Trends'."
)
fig = agent.get_figure()
fig.show()
```
