```python
# Example 3: Customer Satisfaction Survey
import pandas as pd
from plot_agent.agent import PlotAgent


df = pd.DataFrame({
    "response": ["Very Satisfied", "Satisfied", "Neutral", "Dissatisfied", "Very Dissatisfied"],
    "count": [120, 300, 150, 80, 20]
})

agent = PlotAgent()
agent.set_df(df)

# Initial plot
response = agent.process_message(
    "Create a horizontal bar chart of survey response counts, ordering responses from most to least frequent."
)
fig = agent.get_figure()
fig.show()

# Follow-up iteration
response = agent.process_message(
    "Convert it into a pie chart showing percentages, explode the 'Dissatisfied' slice, and add labels with percentage values."
)
fig = agent.get_figure()
fig.show()
```
