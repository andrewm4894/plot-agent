```python
# Example 8: Daily Active Users by Platform
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


df = pd.DataFrame({
    "date": pd.date_range("2025-02-01", periods=60, freq="D").repeat(3),
    "platform": ["iOS", "Android", "Web"] * 60,
    "active_users": np.random.randint(1000, 10000, 180)
})

agent = PlotAgent()
agent.set_df(df)

# Initial plot
response = agent.process_message(
    "Create a multiline chart of daily active users over time, with one line per platform."
)
fig = agent.get_figure()
fig.show()
```

```python
# Follow-up iteration
response = agent.process_message(
    "Add a 7-day rolling average for each line, lighten the weekday lines, and highlight the weekend segments."
)
fig = agent.get_figure()
fig.show()
```
