```python
# Example 6: Geospatial Store Locations
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


df = pd.DataFrame({
    "store": [f"S{i}" for i in range(1, 11)],
    "lat": np.random.uniform(34.0, 42.0, 10),
    "lon": np.random.uniform(-118.0, -70.0, 10),
    "revenue": np.random.randint(50, 500, 10)
})

agent = PlotAgent()
agent.set_df(df)

# Initial plot
response = agent.process_message(
    "Plot store locations on a scatter mapbox, sizing points by revenue."
)
fig = agent.get_figure()
fig.show()

# Follow-up iteration
response = agent.process_message(
    "Color points by revenue quartile, add hover text with store name and exact revenue, and center the map on the data."
)
fig = agent.get_figure()
fig.show()
```
