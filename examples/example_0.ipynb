```python
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


# generate a random time series dataframe with 1000 rows
data = {
    'date': pd.date_range(start='2023-01-01', periods=1000),
    'sales': np.random.randint(100, 1000, size=1000),
    'region': np.random.choice(['North', 'South', 'East', 'West'], size=1000),
    'product': np.random.choice(['A', 'B', 'C'], size=1000)
}

df = pd.DataFrame(data)
print(df.shape)
df.head()
```

```python
# Initialize the agent
agent = PlotAgent()
agent.set_df(df)
```

```python
# Create a line chart showing sales over time, colored by region
response = agent.process_message("Create a line chart showing sales over time, colored by region")
fig = agent.get_figure()
fig.show()
```

```python
# Interactively update the plot
response = agent.process_message("Can you add a title 'Sales Trends by Region' and smooth the lines by making them a 7 day moving average?")
fig = agent.get_figure()
fig.show()
```

```python
# Add a dark mode theme
response = agent.process_message("can you make the plot have a dark mode sort of theme?")
fig = agent.get_figure()
fig.show()
```

```python
# Make a 2x2 sub plot one for each region
response = agent.process_message("can we make a 2x2 sub plot one for each region? make sure to have 4 seperate plots")
fig = agent.get_figure()
fig.show()
```
