```python
# Example 2: Website Traffic by Page & Region
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


df = pd.DataFrame({
    "date": pd.date_range("2025-03-01", periods=30, freq="D").repeat(3),
    "page": ["Home", "Product", "Blog"] * 30,
    "views": np.random.randint(200, 2000, 90),
    "region": np.random.choice(["Americas", "EMEA", "APAC"], 90)
})

agent = PlotAgent()
agent.set_df(df)

# Initial plot
response = agent.process_message(
    "Draw a stacked area chart of daily page views over time, broken out by region."
)
fig = agent.get_figure()
fig.show()

# Follow-up iteration
response = agent.process_message(
    "Switch to a faceted line chart (one subplot per region), add markers at weekends, and include a legend outside the plots."
)
fig = agent.get_figure()
fig.show()
```
