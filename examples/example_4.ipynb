```python
# Example 4: Experiment Response Times
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


df = pd.DataFrame({
    "trial": np.tile(np.arange(1, 6), 40),
    "treatment": np.repeat(["A", "B", "C", "D"], 50),
    "response_ms": np.random.exponential(300, 200) + np.repeat([0, 50, -20, 80], 50)
})

agent = PlotAgent()
agent.set_df(df)

# Initial plot
response = agent.process_message(
    "Draw a box plot of response times (ms) for each treatment group."
)
fig = agent.get_figure()
fig.show()

# Follow-up iteration
response = agent.process_message(
    "Overlay a jittered strip (swarm) plot on the same axes, color points by treatment, and use a log scale for the y-axis."
)
fig = agent.get_figure()
fig.show()
```
