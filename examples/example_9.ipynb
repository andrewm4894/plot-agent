```python
# Example 9: Hierarchical Sales Data (Treemap)
import pandas as pd
import numpy as np
from plot_agent.agent import PlotAgent


df = pd.DataFrame({
    "category": np.random.choice(["Electronics", "Clothing", "Home", "Toys"], 200),
    "sub_category": np.random.choice(["A", "B", "C", "D"], 200),
    "sales": np.random.uniform(100, 2000, 200)
})

agent = PlotAgent()
agent.set_df(df)

# Initial plot
response = agent.process_message(
    "Build a treemap of total sales by category and sub-category."
)
fig = agent.get_figure()
fig.show()

# Follow-up iteration
response = agent.process_message(
    "Use a continuous color scale mapped to sales, add labels showing both category and total, and include a colorbar."
)
fig = agent.get_figure()
fig.show()
```
